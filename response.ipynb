```python
import argparse
import sys

def main():
    parser = argparse.ArgumentParser(
        description="Process a monetary amount with a currency code"
    )

    parser.add_argument(
        "--amount",
        type=float,
        required=True,
        help="Monetary amount (float)"
    )

    parser.add_argument(
        "--currency",
        type=str,
        required=True,
        help="Currency code (e.g., USD, INR, EUR)"
    )

    args = parser.parse_args()

    amount = args.amount
    currency = args.currency

    print(f"Amount: {amount}")
    print(f"Currency: {currency}")

if __name__ == "__main__":
    main()
```
